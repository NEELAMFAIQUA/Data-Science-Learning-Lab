# Explore a small tabular dataset

This original exercise uses fictional practice-time and quiz-score records. Load, inspect, summarize, group, and visualize the data. The example is for learning and does not establish that practice time causes a score change.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

data_path = Path('../data/study_practice.csv')
if not data_path.exists():
    data_path = Path('data/study_practice.csv')
df = pd.read_csv(data_path)
df.head()
```

## Inspect and summarize

Check data types, missing values, and numeric summary statistics before analysis.

```python
df.info()
df.isna().sum(), df.describe(include='all')
```

## Group summary and relationship

Compare group averages and inspect a scatter plot. These are descriptions of a tiny fictional sample, not causal evidence.

```python
df.groupby('group')[['practice_minutes', 'quiz_score']].mean()
```

```python
ax = df.plot.scatter(x='practice_minutes', y='quiz_score', c='tab:blue')
ax.set_title('Fictional practice-time sample')
plt.show()
df[['practice_minutes', 'quiz_score']].corr()
```
